### DRIVERS TABLE

```python
import sqlite3 as sql
import pandas as pd
```

```python
conn = sql.connect('../Data/rideshare.db')
```

```python
drivers_users = pd.read_sql_query("SELECT * FROM drivers LEFT JOIN users ON users.user_id  = drivers.user_id",conn)
drivers_users.to_csv("../src/drivers_users.csv")
drivers_users.head()
```

1. Which city has Highest Number of drivers ?

```python
drivers_users['city'].value_counts()
```

2. Which vehicle is used mostly by drivers ?

```python
drivers_users['vehicle_make'].value_counts()
```

```python
drivers_users['vehicle_model'].value_counts()
```

3. Which vehicle brand has highest rating ?

```python
drivers_users.groupby('vehicle_make')['rating'].mean().sort_values(ascending=False)
```

4. Which vehicle model has highest rating ?

```python
drivers_users.groupby('vehicle_model')['rating'].mean().sort_values(ascending=False)
```

5. Which city has highest driver ratings ?

```python
drivers_users.groupby('city')['rating'].mean().sort_values(ascending=False)
```

6. How many drivers are active ?

```python
drivers_users['is_active'].value_counts()
```
