```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
import sys 
sys.path.append('..')
```

```python
from src.load_data import load_tables
tables = load_tables()
```

```python
users = tables['users']
drivers = tables['drivers']
riders = tables['riders']
locations = tables['locations']
trips = pd.read_csv('../src/cleaned_trips.csv')
payments = tables['payments']
reviews = pd.read_csv('../src/cleaned_reviews.csv')
cancellations = tables['cancellations']
drivers_users = pd.read_csv('../src/drivers_users.csv')
riders_users = pd.read_csv('../src/riders_users.csv')
```

```python
plt.figure(figsize=(8,5),dpi=100)
sns.countplot(data=users,x='city')
plt.title('Number of Users by City',fontdict={'fontname':'monospace','fontsize':20})
plt.savefig('../Images/Numberofusersbycity_countplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5),dpi=100)
sns.countplot(data=drivers_users,x='city')
plt.title('Number of Drivers by City',fontdict={'fontname':'monospace','fontsize':20})
plt.savefig('../Images/Numberofdriversbycity_countplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5),dpi=100)
sns.countplot(data=riders_users,x='city')
plt.title('Number of Riders by City',fontdict={'fontname':'monospace','fontsize':20})
plt.savefig('../Images/Numberofridersbycity_countplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5),dpi=100)
sns.countplot(data=users,x='is_driver')
plt.title('Driver vs Non-Driver User Distribution')
plt.savefig('../Images/DrivervsNonDriver_countplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5),dpi=100)
users['date_joined']=pd.to_datetime(users['date_joined'])
users_byyear = users.groupby(users['date_joined'].dt.year).size()
sns.lineplot(data=users_byyear)
plt.title('Users By Year',fontdict={'fontname':'monospace','fontsize':20})
plt.savefig('../Images/UsersByyear_lineplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5),dpi=100)
users['date_joined']=pd.to_datetime(users['date_joined'])
users_bymonth = users.groupby(users['date_joined'].dt.month).size()
sns.barplot(data=users_bymonth)
plt.title('Users By Month',fontdict={'fontname':'monospace','fontsize':20})
plt.savefig('../Images/UsersBymonth_barplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5),dpi=100)
sns.countplot(data=drivers_users,x='vehicle_make')
plt.title('Mostly Used Vehicle Brand By Drivers')
plt.savefig('../Images/MostlyUsedVehicleBrand_countplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5),dpi=100)
top_5 = drivers_users['vehicle_model'].value_counts().head(5).index
sns.countplot(data=drivers_users,x='vehicle_model',order=top_5)
plt.title('Mostly Used Vehicle Model By Drivers')
plt.savefig('../Images/MostlyUsedVehicleModel_countplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5),dpi=100)
sns.histplot(data=drivers_users,x='rating')
plt.title('Distribution of Driver Rating',fontdict={'fontname':'monospace','fontsize':20})
plt.savefig('../Images/DistofDriverRating_histplot.png',dpi=300)
plt.show()
```
